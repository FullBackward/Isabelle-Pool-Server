## Test sledgehammer

Earlier we invoked sledgehammer by sending the literal `sledgehammer` Isar command as a proof **step** (`execute_command(..., 'sledgehammer')`). Running it that way executes inside the document node and blocks the session's other ML channels (e.g. `open_subgoals`) until it finishes.

Here we use the **standalone sledgehammer channel** instead: `client.sledgehammer(session_id, ...)` hits the dedicated `/sledgehammer` endpoint. On the ML side it forks an Isabelle thread, so it does **not** run as a proof step, does not advance/modify the proof state (read-only), and does not block other sessions. It returns the proof-method suggestions (e.g. `by (metis ...)`) which the prover can then apply as the next step.

## Test session parallelism (`parallel_proofs` / `threads`)

Sessions are now created with configurable PIDE parallelism (`repl/src/main/scala/repl/session_manager.scala`):

- `ISABELLE_PARALLEL_PROOFS` (default `2`) — `2` forks nested `have`/`show` proof bodies so **independent `have`s are checked concurrently**; `1` = top-level proofs only; `0` = fully sequential.
- `ISABELLE_SESSION_THREADS` (default `4`) — caps the ML worker threads per session (deliberately not `0` = auto, so concurrent sessions don't oversubscribe cores and OOM the gateway).

Both are read from the **server's** environment at session creation — set them in `.env` at the repo root and restart the server to change them. Pooled sessions keep the options they were created with.

**The experiment:** one lemma with `N_HAVES` independent `have` steps, each proof sleeping `SLEEP_S` seconds inside an ML tactic before closing the goal. Submitted as **one chunk**, PIDE forks the `have` proofs, so wall time ≈ `ceil(N_HAVES / threads) * SLEEP_S`. The same steps submitted **one command at a time** form the sequential baseline (~`N_HAVES * SLEEP_S`), because each small-step waits for the node to finish checking before returning.

````python
session_payload = await client.acquire_session(["Complex_Main"], "HOL")
session_id = session_payload.get("session_id")
lease_id = session_payload.get("lease_id")
session_payload
````

````python
N_HAVES = 4
SLEEP_S = 2

def slow_have(i: int) -> str:
    # Independent goal; the ML tactic sleeps, then closes it with simp.
    # With parallel_proofs=2 these `by` bodies are forked and run concurrently.
    return (
        f'have h{i}: "{i} + 0 = ({i}::nat)"\n'
        f"  by (tactic ‹(OS.Process.sleep (Time.fromSeconds {SLEEP_S}); simp_tac @{{context}} 1)›)"
    )

parallel_chunk = "\n".join(
    ["lemma par_demo: True", "proof -"]
    + [slow_have(i) for i in range(1, N_HAVES + 1)]
    + ["show True by simp", "qed"]
)
print(parallel_chunk)
````

````python
await client.enter_theory(session_id, "Par_Demo", lease_id=lease_id)
await client.execute_command(session_id, "theory Par_Demo imports Complex_Main begin", lease_id=lease_id)
````

````python
import time

t0 = time.perf_counter()
res = await client.execute_command(session_id, parallel_chunk, timeout=120.0, lease_id=lease_id)
parallel_wall = time.perf_counter() - t0
print(f"one chunk, {N_HAVES} parallel haves: {parallel_wall:.1f}s "
      f"(sequential would be ~{N_HAVES * SLEEP_S}s of sleep alone)")
res
````

````python
# Sequential baseline: same steps, one command per request.
# Each step waits for checking to finish, so the sleeps cannot overlap.
t0 = time.perf_counter()
await client.execute_command(session_id, "lemma seq_demo: True", timeout=120.0, lease_id=lease_id)
await client.execute_command(session_id, "proof -", timeout=120.0, lease_id=lease_id)
for i in range(1, N_HAVES + 1):
    await client.execute_command(session_id, slow_have(i), timeout=120.0, lease_id=lease_id)
await client.execute_command(session_id, "show True by simp", timeout=120.0, lease_id=lease_id)
await client.execute_command(session_id, "qed", timeout=120.0, lease_id=lease_id)
sequential_wall = time.perf_counter() - t0
print(f"step-by-step: {sequential_wall:.1f}s | one chunk: {parallel_wall:.1f}s "
      f"| speedup x{sequential_wall / parallel_wall:.1f}")
````

````python
await client.execute_command(session_id, "end", lease_id=lease_id)
await client.release_session(session_id, lease_id=lease_id)
````

To see the knobs bite, edit `.env`, restart the server, and rerun the chunk cell (the pool is rebuilt on restart, so new sessions pick up the options):

| `.env` setting | expected chunk wall time |
|---|---|
| `ISABELLE_PARALLEL_PROOFS=2`, `ISABELLE_SESSION_THREADS=4` | ≈ `SLEEP_S` (all 4 haves in flight) |
| `ISABELLE_PARALLEL_PROOFS=2`, `ISABELLE_SESSION_THREADS=2` | ≈ `2 × SLEEP_S` (two at a time) |
| `ISABELLE_PARALLEL_PROOFS=0` | ≈ `N_HAVES × SLEEP_S` (no forking — matches the step-by-step baseline) |

| `parallel_proofs=0` (forking off) | one chunk, same 4 `have`s | **24.9 s** | 24.72 s |

Takeaways:

- With forking on, the chunk runs **×3.8** faster than the step-by-step baseline (26.6 s → 6.9 s) — the speedup matches `N_HAVES = 4`, i.e. all four `have` bodies were checked concurrently.
- With `parallel_proofs=0` the *same chunk* degrades to 24.9 s ≈ `N_HAVES × SLEEP_S` + overhead — indistinguishable from the step-by-step baseline, confirming the speedup comes from PIDE proof forking and not from chunked submission itself.
- The parallel chunk does not hit the theoretical ~`SLEEP_S` floor (2 s) because the amd64 image runs under qemu emulation on this host, which inflates per-step checking overhead (visible also in the 26.6 s baseline vs ~8 s of pure sleep).

## Big-chunk verification with `verify_chunk`

The sledgehammer section above advanced the `amc12b_2020_p22` proof **one command per call** — many round-trips, and any single line could time out on its own with no indication of *which* line hung.

`verify_chunk` submits the **whole proof body in one call**. It is checked with per-session parallelism on, bounded by a **single** wall budget, and returns a **per-command status report** in source order — each command is `ok`, `failed`, `running`, or `unprocessed`. On a timeout the report is *partial* and names the still-`running` line (the likely loop) instead of surfacing an opaque timeout.

Below we verify the **exact same proof** from the sledgehammer section as a single chunk.

````python
# verify_chunk needs the theory already begun, so acquire a session, enter the
# theory, and process the header as a normal step first.
chunk_payload = await client.acquire_session(["Complex_Main"], "HOL")
chunk_sid = chunk_payload["session_id"]
chunk_lease = chunk_payload["lease_id"]
await client.enter_theory(chunk_sid, "amc12b_2020_p22", lease_id=chunk_lease)
await client.execute_command(
    chunk_sid,
    'theory amc12b_2020_p22 imports Complex_Main "HOL-Library.Code_Target_Numeral" begin',
    lease_id=chunk_lease,
)
````

````python
# The same theorem..qed proof from the sledgehammer section, as ONE chunk.
proof_chunk = r"""theorem amc12b_2020_p22: fixes t :: real shows "((2 powr t - 3 * t) * t) / (4 powr t) ≤ 1 / 12"
proof -
have hpow: "(2 powr t)^2 = 4 powr t"
by (metis powr_mult numeral_Bit0_eq_double power2_eq_square)
have pos: "(0::real) < 4 powr t" by simp
have amgm: "2 * (2 powr t) * (6 * t) ≤ (2 powr t)^2 + (6 * t)^2"
by (rule sum_squares_bound)
have rearr: "12 * ((2 powr t - 3 * t) * t) = 2 * (2 powr t) * (6 * t) - (6 * t)^2"
by (simp add: power2_eq_square algebra_simps)
from amgm rearr hpow have key: "12 * ((2 powr t - 3 * t) * t) ≤ 4 powr t"
by linarith
from key pos show ?thesis
by (simp add: field_simps)
qed
"""

# Single call, single wall budget; per-command status comes back in source order.
report = await client.verify_chunk(chunk_sid, proof_chunk, timeout=180.0, lease_id=chunk_lease)

# Built-in pretty printer in the async client — no custom loop needed.
client.print_chunk_report(report)
report
````

````python
# Finish the theory and return the session to the pool.
await client.execute_command(chunk_sid, 'end', timeout=120.0, lease_id=chunk_lease)
await client.close_session(chunk_sid, lease_id=chunk_lease)
````

## Diagnostic queries with `diagnostic`

`verify_chunk` reports only whether each command is `ok`/`failed` (plus any error/warning text) — it **drops the `writeln` output** that *diagnostic* commands produce. So asking `verify_chunk` to run `thm conjI` tells you the command succeeded but **not what the theorem says**.

The `diagnostic` endpoint fills that gap. It runs a single **read-only** Isabelle query — `thm`, `term`, `prop`, `typ`, `prf`, `find_theorems`, `find_consts`, any `print_*`/`find_*` — **transiently** (insert → capture output → discard), so it returns the query's output **without changing the proof script or the rollback chain**. Input is gatekept by an allowlist of diagnostic keywords plus a denylist of code-executing / IO commands (`ML`, `setup`, `*_file`, …), which are rejected with HTTP 422.

It is exposed as `client.diagnostic(session_id, command, lease_id=...)` and, for agents, as the MCP `diagnostic` tool. The session must already have a theory begun (via `enter_theory`).

### Contrast with `verify_chunk`, transience, and the safety gate

Running the *same* `thm conjI` through `verify_chunk` confirms the command succeeds but returns **no message text** — the writeln output is not part of the proof-status report. That is exactly why `diagnostic` exists.

`diagnostic` is also **transient** (the theory source is unchanged after running queries) and **gatekept** (a code-executing or state-mutating command is rejected with HTTP 422 before it ever reaches the prover).

## LSP-style read-only mode (file sync, per-line diagnostics, position queries)

New server surface for editor/agent-style interaction (design: `claude-work/2026-8-8-research-lsp-readonly-mode/`). The idea: **the agent edits a `.thy` file on disk; the session mirrors it** — `load_document` replaces the session's whole document with the file text (disk is the source of truth), and every query is read-only.

Features shown below:
- `load_document(..., report=True)` — whole-file load with a **per-command status report with line+column ranges** (failures are *not* rolled back, so broken state stays inspectable, LSP-style).
- `goals_at_line` / `command_at_line` — jEdit-style position queries, snapshot-based (zero document mutation).
- `get_local_facts` / `get_global_facts`, `get_last_report` — retained diagnostics.
- Queries on **finished theories** (trailing `end`) — no more probe hangs.

````python
# A demo .thy file ON DISK with a deliberate error on line 4.
demo_file = "lsp_demo.thy"
with open(demo_file, "w") as f:
    f.write("""theory LspDemo imports Main begin

lemma ok1: True by simp
lemma bad: False by auto
lemma ok2: True by simp
""")

# Create a session and mirror the file into it (full-file mode: the text carries its own header).
lsp = await client.create_session(["Main"], "HOL", label="demo/lsp_demo.thy")
lsp_sid, lsp_lease = lsp["session_id"], lsp["lease_id"]

with open(demo_file) as f:
    text = f.read()
resp = await client.load_document(lsp_sid, text, timeout=120, report=True, lease_id=lsp_lease)
print("load success:", resp["success"], "| theory:", resp["theory"], "| error:", resp["error"])
print()
print(client.format_chunk_report(resp["report"]))
````

````python
# The report carries a per-command RANGE (1-based line/col, UTF-16 columns) —
# the failing command's span maps directly onto the editor buffer.
for c in resp["report"]["commands"]:
    if c["status"] == "failed":
        r = c["range"]
        print(f'FAILED: kind={c["kind"]} at lines {r["start"]["line"]}:{r["start"]["col"]}–{r["end"]["line"]}:{r["end"]["col"]}')
        print("message:", c["messages"][0]["text"][:120])
# Note: failures are NOT rolled back — ok2 (after the error) still checked, and ok1 is a live fact:
(await client.diagnostic(lsp_sid, "thm ok1", lease_id=lsp_lease))["output"]
````

````python
# The file-sync model: edit the file ON DISK, reload — the session follows the file.
with open(demo_file, "w") as f:
    f.write("""theory LspDemo imports Main begin

lemma ok1: True by simp
lemma bad: False by simp  (* still wrong, but now it fails differently *)
""")

with open(demo_file) as f:
    text2 = f.read()
resp2 = await client.load_document(lsp_sid, text2, timeout=120, report=True, lease_id=lsp_lease)
print("reload success:", resp2["success"], "| error:", resp2["error"])

# The last report is retained server-side — queryable later without re-running anything:
last = await client.get_last_report(lsp_sid, lease_id=lsp_lease)
print("last_report commands:", [(c["kind"], c["status"]) for c in last["report"]["commands"]])
````

### Position queries (snapshot-based, read-only)

`goals_at_line` returns the proof state **before/after** the command containing a line (like `lean_goal`); `command_at_line` returns the enclosing command with its range. Both read retained per-command results from an immutable snapshot — **no probes, no document mutation**, and they work past a trailing `end`.

````python
proof_text = """theory GoalDemo imports Main begin
lemma conj_swap: "A ∧ B ⟹ B ∧ A"
proof (rule conjI)
  assume "A ∧ B"
  show B by (rule conjunct2[OF ‹A ∧ B›])
  show A by (rule conjunct1[OF ‹A ∧ B›])
qed
"""
await client.load_document(lsp_sid, proof_text, timeout=120, lease_id=lsp_lease)

for line in [2, 3, 5]:
    g = await client.goals_at_line(lsp_sid, line, lease_id=lsp_lease)
    print(f"--- line {line}: {g['command']['kind']!r:12} range={g['command']['range']['start']['line']}")
    print("  before:", (g["goals_before"] or ["(none)"])[0][:80].replace("\n", " ⏎ "))
    print("  after: ", (g["goals_after"] or ["(none)"])[0][:80].replace("\n", " ⏎ "))

# A comment/blank line resolves to the enclosing command (ignored-span aware):
cmd = await client.command_at_line(lsp_sid, 4, lease_id=lsp_lease)
print("line 4 (assume):", cmd["kind"], "->", cmd["source"][:50])
````

````python
# Fact queries mid-proof (transient probes): local context vs theory-level facts.
await client.load_document(
    lsp_sid,
    'theory FactDemo imports Main begin\nlemma "⟦ A; B ⟧ ⟹ A ∧ B"\nproof -\n  assume A B',
    timeout=120, lease_id=lsp_lease,
)
print("local facts:", (await client.get_local_facts(lsp_sid, lease_id=lsp_lease))["facts"])
print()
glob = await client.get_global_facts(lsp_sid, limit=5, lease_id=lsp_lease)
print(f"global facts (first {glob['count']} of Main):")
for f in glob["facts"]:
    print("  ", f[:70])
````

````python
# Finished theories: after a trailing `end`, queries used to hang (2× channel timeout → 500).
# Now they answer immediately — probes insert BEFORE the `end`; subgoals/in_proof short-circuit.
finished = """theory DoneDemo imports Main begin
lemma foo: True by simp
end
"""
await client.load_document(lsp_sid, finished, timeout=120, report=True, lease_id=lsp_lease)

import time
t0 = time.time()
out = (await client.diagnostic(lsp_sid, "thm foo", lease_id=lsp_lease))["output"]
subs = await client.get_subgoals(lsp_sid, lease_id=lsp_lease)
print(f"thm foo -> {out!r}  ({time.time()-t0:.2f}s)")
print("subgoals after end:", subs["subgoals"], "| proof_finished:", subs["proof_finished"])
````

## LSP-style MCP server (`mcp_lsp_server`) — full tool tour

Everything above drove the HTTP API directly. This section talks to the **MCP server** the way an LLM agent harness does: it spawns `python -m mcp_lsp_server.app` as a stdio subprocess and calls its tools over the MCP protocol — no HTTP client involved.

The workflow is the lean-lsp-mcp file-sync model: tools take a `file_path` (never a session id), the MCP re-reads the file from disk before every call and re-syncs the session (`load_document(report=true)`), and execution tools (`isabelle_multi_attempt`, `isabelle_run_code`) run on a warm **scratch** pool — never on the file's session. The MCP never writes files; the notebook writes the demo file itself.

Prerequisites: the gym HTTP server on `:8000` (as for all sections above) and `pip install "mcp<2"` in the notebook's environment (mcp 2.0 removed `mcp.server.fastmcp`).


````python
import os, sys, json
from contextlib import AsyncExitStack
from mcp import ClientSession
from mcp.client.stdio import StdioServerParameters, stdio_client

MCP_DEMO = os.path.join(REPO_ROOT, "demo", "mcp_lsp_demo.thy")

# v1: a deliberate error on line 4, for the diagnostics demo.
V1 = 'theory McpLspDemo imports Main begin\n\nlemma ok1: True by simp\nlemma broken: False by simp\n\nend\n'
with open(MCP_DEMO, "w") as f:
    f.write(V1)

async def mcp_call(name, **args):
    """Call an MCP tool and print its text payload (tools return JSON strings)."""
    result = await mcp_sess.call_tool(name, args)
    text = "\n".join(c.text for c in result.content if getattr(c, "type", "") == "text")
    print(f"--- {name} ---")
    print(text[:1500])
    print()
    return text
````

````python
# Spawn the MCP server (host-side subprocess) and connect over stdio.
# PYTHONPATH must include the repo root so `mcp_lsp_server` and `client` import.
stack = AsyncExitStack()
env = dict(os.environ, PYTHONPATH=REPO_ROOT + os.pathsep + os.environ.get("PYTHONPATH", ""))
params = StdioServerParameters(command=sys.executable, args=["-m", "mcp_lsp_server.app"], env=env)
read, write = await stack.enter_async_context(stdio_client(params))
mcp_sess = await stack.enter_async_context(ClientSession(read, write))
await mcp_sess.initialize()

tools = await mcp_sess.list_tools()
print(f"{len(tools.tools)} tools:")
print(", ".join(sorted(t.name for t in tools.tools)))
````

````python
# Open the file (binds a session + first sync) and read the diagnostics:
# the broken lemma shows up with its 1-based line/col range (UTF-16 columns).
await mcp_call("isabelle_open", file_path=MCP_DEMO)
await mcp_call("isabelle_diagnostic_messages", file_path=MCP_DEMO)
````

````python
# The file-sync model: fix the file ON DISK, then sync — the session follows the file.
V2 = 'theory McpLspDemo imports Main begin\n\nlemma hd_id: "hd [x] = x" by simp\nlemma use_hd: "hd [1::nat] = 1"\n  by (simp add: hd_id)\n\nlemma two_goals: "\\<lbrakk> A; B \\<rbrakk> \\<Longrightarrow> A \\<and> B"\nproof (rule conjI)\n  show A by assumption\n  show B by assumption\nqed\n\nend\n'
with open(MCP_DEMO, "w") as f:
    f.write(V2)
await mcp_call("isabelle_sync", file_path=MCP_DEMO)
await mcp_call("isabelle_diagnostic_messages", file_path=MCP_DEMO)  # clean now
````

````python
# Position queries (snapshot-based, read-only; 1-based lines).
# Line 8 is `proof (rule conjI)` — after it, two subgoals are open.
await mcp_call("isabelle_goal", file_path=MCP_DEMO, line=8)
await mcp_call("isabelle_command_at_line", file_path=MCP_DEMO, line=8)
await mcp_call("isabelle_proof_state", file_path=MCP_DEMO)   # tip state (no open proof)
src = await mcp_call("isabelle_source", file_path=MCP_DEMO)  # source as the session sees it
````

````python
# Transient read-only probes (never touch the proof script):
# thm/term/find_theorems-style queries, and local vs theory-level fact lists.
await mcp_call("isabelle_query", file_path=MCP_DEMO, command="thm hd_id")
await mcp_call("isabelle_local_facts", file_path=MCP_DEMO)
await mcp_call("isabelle_global_facts", file_path=MCP_DEMO, limit=5)
````

````python
# hover / go-to-definition at a position (columns computed from the text — no magic numbers).
line_no = 3
line_text = V2.splitlines()[line_no - 1]
col_hd = line_text.index('"hd') + 2      # the library constant `hd`
col_own = line_text.index("hd_id") + 1   # our own lemma `hd_id`
await mcp_call("isabelle_hover_info", file_path=MCP_DEMO, line=line_no, column=col_hd)
await mcp_call("isabelle_definition", file_path=MCP_DEMO, line=line_no, column=col_hd)
await mcp_call("isabelle_definition", file_path=MCP_DEMO, line=line_no, column=col_own)
````

````python
# Heap tools: build a verified heap for the demo project (LONG-RUNNING — isabelle
# build -b), inspect its manifest, then open a consumer file bound to that heap.
# The project path must be visible to the SERVER — the repo is mounted at /app in
# the container. The consumer file itself is read from disk by the MCP (host path).
build = await mcp_call("isabelle_build_heap", task_group="demo-mcp",
                       project="/app/demo/heap_demo_project", session_name="McpDemoHeap")
await mcp_call("isabelle_heap_status", task_group="demo-mcp")

CONSUMER = 'theory McpHeapConsumer imports "McpDemoHeap.Shared" begin\nlemma use_it: True by (simp add: shared_lemma)\nend\n'
CONSUMER_FILE = os.path.join(REPO_ROOT, "demo", "mcp_heap_consumer.thy")
with open(CONSUMER_FILE, "w") as f:
    f.write(CONSUMER)
await mcp_call("isabelle_open", file_path=CONSUMER_FILE,
               task_group="demo-mcp", heap_session="McpDemoHeap")
await mcp_call("isabelle_query", file_path=CONSUMER_FILE, command="thm shared_lemma")
````

````python
# Cleanup: unbind the files (sessions go back to the server pool for warm reuse).
# The demo-mcp heap persists — delete it via GET /admin or the HTTP heap
# endpoints if you don't want it.
await mcp_call("isabelle_close", file_path=CONSUMER_FILE)
await mcp_call("isabelle_close", file_path=MCP_DEMO)
# Graceful aclose() crosses Jupyter's per-cell task boundary — anyio cancel
# scopes are task-bound, so exiting a stack entered in another cell raises
# "Attempted to exit cancel scope in a different task". The sessions are
# already released above, so fall back to terminating the MCP subprocess.
try:
    await stack.aclose()
except BaseException as e:
    print(f"graceful aclose unavailable across cells ({type(e).__name__}); terminating subprocess")
    import subprocess as sp
    for pid in sp.check_output(["pgrep", "-f", "mcp_lsp_server.app"]).split():
        os.kill(int(pid), 15)
print("MCP server shut down.")
````
